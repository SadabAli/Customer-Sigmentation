## Feature Preparation for Customer Clustering

```python

# Inspect the columns before selecting clustering features
print("Dataset shape:", df.shape)

print("\nUnique values in important columns:")
for col in [
    "CustomerID",
    "Age",
    "Gender",
    "Category",
    "PurchaseAmountUSD",
    "Season",
    "PaymentMethod",
    "FrequencyOfPurchases"
]:
    print(f"{col}: {df[col].nunique()} unique values")

print("\nCustomer ID sample:")
display(df[["CustomerID"]].head(10))

print("\nDataset columns:")
print(df.columns.tolist())
```

### Select clustering features

```python

# Create a separate dataset for clustering
# Keep the original df unchanged

cluster_features = [
    "Age",
    "Gender",
    "Category",
    "PurchaseAmountUSD",
    "Season",
    "ReviewRating",
    "SubscriptionStatus",
    "DiscountApplied",
    "PromoCodeUsed",
    "PreviousPurchases",
    "PaymentMethod",
    "FrequencyOfPurchases"
]

X = df[cluster_features].copy()

print("Clustering feature shape:", X.shape)
print("\nData types:")
display(X.dtypes.to_frame("Data Type"))

print("\nMissing values in clustering features:")
display(X.isnull().sum().to_frame("Missing Values"))

print("\nSample clustering records:")
display(X.head())
```

### Why these features?
- Age and purchase amount: represent shopper demographics and spending.
- Category and season: represent the type and timing of shopping.
- Review rating: represents the rating associated with a purchase.
- Subscription, discount, and promo code: describe shopping and promotion-related attributes.
- Previous purchases and frequency: provide available indicators of purchasing activity.
- Payment method: describes how the purchase was paid for.

We are leaving out CustomerID because it is just an identifier, and AgeGroup because it duplicates information already represented by Age. We'll also leave out ItemPurchased, Location, Size, Color, and ShippingType in this first experiment to keep the feature set manageable. We can test additional features later.

Important limitation: because each customer ID appears only once, these clusters represent shopping-record profiles, not verified groups of repeat customers. PreviousPurchases is still available as a field, but we cannot link those purchases to other rows for the same customer.

### Prepare the clustering data

```python

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
import pandas as pd

# Separate numerical and categorical features
numeric_features = [
    "Age",
    "PurchaseAmountUSD",
    "ReviewRating",
    "PreviousPurchases"
]

categorical_features = [
    "Gender",
    "Category",
    "Season",
    "SubscriptionStatus",
    "DiscountApplied",
    "PromoCodeUsed",
    "PaymentMethod",
    "FrequencyOfPurchases"
]

# Preprocess numeric and categorical columns
preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numeric_features),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features)
    ]
)

# Fit and transform the selected features
X_prepared_array = preprocessor.fit_transform(X)

# Convert the processed result into a DataFrame
feature_names = preprocessor.get_feature_names_out()

X_prepared = pd.DataFrame(
    X_prepared_array.toarray()
    if hasattr(X_prepared_array, "toarray")
    else X_prepared_array,
    columns=feature_names
)

print("Prepared data shape:", X_prepared.shape)
print("\nMissing values:", X_prepared.isnull().sum().sum())
print("\nFirst five prepared records:")
display(X_prepared.head())

# Save the prepared data for Orange
X_prepared.to_csv("customer_segmentation_prepared.csv", index=False)
print("\nSaved: customer_segmentation_prepared.csv")
```

### Why are we doing this?
- StandardScaler: puts numerical features on comparable scales.
- OneHotEncoder: converts categories such as Gender, Season, and PaymentMethod into numeric indicator columns.
- CustomerID excluded: prevents a unique identifier from distorting similarity calculations.
- Original dataset preserved: we can still use the original columns to interpret the clusters later.

Important: this creates a prepared dataset for an initial clustering experiment. One-hot encoding can create many columns, and the resulting groups must still be evaluated for quality and usefulness.

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.neighbors import NearestNeighbors
from sklearn.cluster import DBSCAN
from sklearn.metrics import silhouette_score

# Use the 29 prepared features
X_dbscan = X_prepared.to_numpy(dtype=float)

# Estimate distances to the 5th nearest neighbor
min_samples_reference = 5

neighbors = NearestNeighbors(
    n_neighbors=min_samples_reference
)
neighbors.fit(X_dbscan)

distances, _ = neighbors.kneighbors(X_dbscan)
k_distances = np.sort(distances[:, -1])

plt.figure(figsize=(10, 5))
plt.plot(k_distances)
plt.title("DBSCAN k-Distance Plot (k=5)")
plt.xlabel("Records sorted by distance")
plt.ylabel("Distance to 5th nearest neighbor")
plt.tight_layout()
plt.show()

print("Suggested eps search values:")
for q in [0.70, 0.80, 0.90, 0.95, 0.98]:
    print(f"{q:.0%} percentile: {np.quantile(k_distances, q):.4f}")
```

```python

# Candidate eps values based on the k-distance distribution
eps_values = [
    float(np.quantile(k_distances, q))
    for q in [0.70, 0.80, 0.90, 0.95, 0.98]
]

min_samples_values = [5, 10, 15, 20]

results = []

for eps in eps_values:
    for min_samples in min_samples_values:

        model = DBSCAN(
            eps=eps,
            min_samples=min_samples,
            metric="euclidean"
        )

        labels = model.fit_predict(X_dbscan)

        # -1 represents noise, not a regular cluster
        non_noise = labels != -1
        cluster_labels = set(labels[non_noise])
        n_clusters = len(cluster_labels)
        noise_pct = (~non_noise).mean() * 100

        score = np.nan
        n_non_noise = non_noise.sum()

        # Silhouette is meaningful only with 2+ non-noise clusters
        if n_clusters >= 2 and n_non_noise > n_clusters:
            score = silhouette_score(
                X_dbscan[non_noise],
                labels[non_noise],
                sample_size=min(1000, n_non_noise),
                random_state=42
            )

        results.append({
            "eps": round(eps, 4),
            "min_samples": min_samples,
            "clusters_excluding_noise": n_clusters,
            "noise_pct": round(noise_pct, 2),
            "silhouette_excluding_noise": score
        })

dbscan_results = pd.DataFrame(results)

display(
    dbscan_results.sort_values(
        "silhouette_excluding_noise",
        ascending=False,
        na_position="last"
    )
)
```

```python

from sklearn.cluster import AgglomerativeClustering
from sklearn.metrics import silhouette_score
import pandas as pd

X_hier = X_prepared.to_numpy(dtype=float)

results = []

for k in range(2, 11):
    model = AgglomerativeClustering(
        n_clusters=k,
        linkage="ward"
    )

    labels = model.fit_predict(X_hier)

    score = silhouette_score(
        X_hier,
        labels,
        sample_size=1000,
        random_state=42
    )

    sizes = pd.Series(labels).value_counts()

    results.append({
        "Clusters": k,
        "Silhouette": round(score, 4),
        "Smallest_Cluster": int(sizes.min()),
        "Largest_Cluster": int(sizes.max())
    })

hier_results = pd.DataFrame(results)

display(
    hier_results.sort_values(
        "Silhouette",
        ascending=False
    )
)
```

```python

from sklearn.mixture import GaussianMixture
from sklearn.metrics import silhouette_score
import pandas as pd

X_gmm = X_prepared.to_numpy(dtype=float)

gmm_results = []
gmm_models = {}

for k in range(2, 11):
    model = GaussianMixture(
        n_components=k,
        covariance_type="diag",
        n_init=3,
        max_iter=200,
        random_state=42
    )

    labels = model.fit_predict(X_gmm)
    score = silhouette_score(
        X_gmm,
        labels,
        sample_size=1000,
        random_state=42
    )

    sizes = pd.Series(labels).value_counts()

    gmm_models[k] = model

    gmm_results.append({
        "Clusters": k,
        "Silhouette": round(score, 4),
        "Smallest_Cluster": int(sizes.min()),
        "Largest_Cluster": int(sizes.max()),
        "BIC": round(model.bic(X_gmm), 2)
    })

gmm_results_df = pd.DataFrame(gmm_results)

display(
    gmm_results_df.sort_values(
        "Silhouette",
        ascending=False
    )
)

```

```python

import pandas as pd

df = pd.read_csv("customer_segmentation_data.csv")

print("Dataset shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

display(df.head())
```

```python

import numpy as np
from kmodes.kprototypes import KPrototypes

# Numerical features
numeric_cols = [
    "Age",
    "PurchaseAmountUSD",
    "PreviousPurchases"
]

# Categorical features
categorical_cols = [
    "Category",
    "SubscriptionStatus",
    "DiscountApplied",
    "PromoCodeUsed",
    "FrequencyOfPurchases"
]

feature_cols = numeric_cols + categorical_cols

X = df[feature_cols].copy()

# Ensure correct data types
for col in numeric_cols:
    X[col] = pd.to_numeric(X[col], errors="raise")

for col in categorical_cols:
    X[col] = X[col].astype(str)

# K-Prototypes expects categorical columns as object values
X = X.astype({
    col: "object" for col in categorical_cols
})

# Numerical features first, categorical features after
X_array = X.to_numpy(dtype=object)
categorical_indices = list(
    range(len(numeric_cols), len(feature_cols))
)

print("Rows:", X_array.shape[0])
print("Features:", X_array.shape[1])
print("Categorical column indices:", categorical_indices)
print("Missing values:", X.isna().sum().sum())
```

```python

kproto = KPrototypes(
    n_clusters=3,
    init="Cao",
    n_init=5,
    max_iter=100,
    random_state=42,
    verbose=0
)

clusters = kproto.fit_predict(
    X_array,
    categorical=categorical_indices
)

df["Cluster"] = clusters

print("Cluster sizes:")
print(df["Cluster"].value_counts().sort_index())

print("\nCluster percentages:")
print(
    (df["Cluster"].value_counts(normalize=True).sort_index() * 100)
    .round(2)
)

print("\nCost:", round(kproto.cost_, 2))
```

```python

# Numerical profiles
print("NUMERICAL PROFILE")
display(
    df.groupby("Cluster")[
        ["Age", "PurchaseAmountUSD", "PreviousPurchases"]
    ].mean().round(2)
)

# Categorical profiles
categorical_profile_cols = [
    "Category",
    "SubscriptionStatus",
    "DiscountApplied",
    "PromoCodeUsed",
    "FrequencyOfPurchases"
]

for col in categorical_profile_cols:
    print(f"\nMost common {col} in each cluster:")
    print(
        df.groupby("Cluster")[col]
        .agg(lambda s: s.mode().iloc[0])
    )
```

```python

from sklearn.preprocessing import StandardScaler

# Keep the original dataframe unchanged for interpretation
X_scaled = X.copy()

# Standardize numerical features
scaler = StandardScaler()
X_scaled[numeric_cols] = scaler.fit_transform(X[numeric_cols])

# Preserve categorical values as strings
for col in categorical_cols:
    X_scaled[col] = X_scaled[col].astype(str)

X_scaled_array = X_scaled.to_numpy(dtype=object)

# Fit a new K-Prototypes model with K=3
kproto_scaled = KPrototypes(
    n_clusters=3,
    init="Cao",
    n_init=5,
    max_iter=100,
    random_state=42,
    verbose=0
)

scaled_clusters = kproto_scaled.fit_predict(
    X_scaled_array,
    categorical=categorical_indices
)

# Save the new labels separately
df["Cluster_Scaled"] = scaled_clusters

print("Standardized model cluster sizes:")
print(df["Cluster_Scaled"].value_counts().sort_index())

print("\nStandardized model cost:", round(kproto_scaled.cost_, 2))

print("\nAverage purchase amount by cluster:")
display(
    df.groupby("Cluster_Scaled")["PurchaseAmountUSD"]
    .agg(["count", "mean", "median"])
    .round(2)
)
```

```python

# Numerical profile in original units
print("NUMERICAL PROFILE")
display(
    df.groupby("Cluster_Scaled")[
        ["Age", "PurchaseAmountUSD", "PreviousPurchases"]
    ].mean().round(2)
)

# Categorical profile: most common value and its percentage
for col in categorical_cols:
    print(f"\n{col}")
    profile = df.groupby("Cluster_Scaled")[col].agg(
        lambda s: f"{s.value_counts().index[0]} "
                  f"({s.value_counts(normalize=True).iloc[0] * 100:.1f}%)"
    )
    print(profile)

# Check whether the two higher-purchase clusters differ
print("\nMEDIAN PURCHASE AMOUNT")
display(
    df.groupby("Cluster_Scaled")["PurchaseAmountUSD"]
    .median()
    .round(2)
)
```

```python
summary = df.groupby("Cluster_Scaled").agg(
    Customers=("CustomerID", "count"),
    Avg_Age=("Age", "mean"),
    Avg_Purchase_Amount=("PurchaseAmountUSD", "mean"),
    Median_Purchase_Amount=("PurchaseAmountUSD", "median"),
    Avg_Previous_Purchases=("PreviousPurchases", "mean")
).round(2)

summary["Customer_Percentage"] = (
    summary["Customers"] / len(df) * 100
).round(1)

print("FINAL CUSTOMER SEGMENT SUMMARY")
display(summary)

summary.to_csv("customer_segment_summary.csv")
```

```python

!pip -q install kmodes gower

import numpy as np
import pandas as pd
import gower
from kmodes.kprototypes import KPrototypes
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score

numeric_cols = ["Age", "PurchaseAmountUSD", "PreviousPurchases"]
categorical_cols = [
    "Category", "SubscriptionStatus", "DiscountApplied",
    "PromoCodeUsed", "FrequencyOfPurchases"
]
feature_cols = numeric_cols + categorical_cols

# Prepare the same mixed features for every experiment
X = df[feature_cols].copy()
scaler = StandardScaler()
X[numeric_cols] = scaler.fit_transform(X[numeric_cols])

for col in categorical_cols:
    X[col] = X[col].astype(str)

X_array = X.to_numpy(dtype=object)
cat_idx = list(range(len(numeric_cols), len(feature_cols)))

# Use one fixed sample for a fair comparison
rng = np.random.default_rng(42)
idx = rng.choice(len(df), size=min(600, len(df)), replace=False)

X_raw_sample = df[feature_cols].iloc[idx].reset_index(drop=True)
D = gower.gower_matrix(X_raw_sample)

results = []

for k in [2, 3, 4]:
    model = KPrototypes(
        n_clusters=k, init="Cao", n_init=5,
        max_iter=100, random_state=42
    )
    labels = model.fit_predict(X_array, categorical=cat_idx)

    score = silhouette_score(
        D, labels[idx], metric="precomputed"
    )

    results.append({
        "K": k,
        "Gower_Silhouette": round(score, 4),
        "Cluster_Sizes": np.bincount(labels).tolist(),
        "Model_Cost": round(model.cost_, 2)
    })

display(pd.DataFrame(results))
```

```python

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.decomposition import PCA
from sklearn.pipeline import make_pipeline

# 1. Select the same eight features used for clustering
numeric_cols = [
    "Age", "PurchaseAmountUSD", "PreviousPurchases"
]

categorical_cols = [
    "Category", "SubscriptionStatus", "DiscountApplied",
    "PromoCodeUsed", "FrequencyOfPurchases"
]

feature_cols = numeric_cols + categorical_cols
X_vis = df[feature_cols].copy()

# 2. Scale numerical features and encode categorical features
preprocessor = ColumnTransformer([
    ("num", StandardScaler(), numeric_cols),
    ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_cols)
])

X_encoded = preprocessor.fit_transform(X_vis)

# 3. Reduce the features to two dimensions for plotting
pca = PCA(n_components=2, random_state=42)
coordinates = pca.fit_transform(X_encoded)

plot_df = pd.DataFrame({
    "PC1": coordinates[:, 0],
    "PC2": coordinates[:, 1],
    "Cluster": df["Cluster_Scaled"].astype(str).to_numpy()
})

# 4. Plot the three clusters
plt.figure(figsize=(10, 7))

for cluster in sorted(plot_df["Cluster"].unique()):
    subset = plot_df[plot_df["Cluster"] == cluster]
    plt.scatter(
        subset["PC1"],
        subset["PC2"],
        label=f"Cluster {cluster}",
        alpha=0.6,
        s=20
    )

plt.title("Customer Segmentation: 3 Clusters (PCA Visualization)")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.legend()
plt.grid(alpha=0.2)
plt.tight_layout()
plt.show()

print(
    "Variance explained by 2D plot:",
    round(pca.explained_variance_ratio_.sum() * 100, 2),
    "%"
)
```

```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    f1_score
)

# Input features and pseudo-label target
numeric_cols = [
    "Age", "PurchaseAmountUSD", "PreviousPurchases"
]
categorical_cols = [
    "Category", "SubscriptionStatus", "DiscountApplied",
    "PromoCodeUsed", "FrequencyOfPurchases"
]

X = df[numeric_cols + categorical_cols].copy()
y = df["Cluster_Scaled"]

# Split data before fitting the preprocessing/model pipeline
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

preprocessor = ColumnTransformer([
    ("num", StandardScaler(), numeric_cols),
    ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_cols)
])

model = Pipeline([
    ("preprocessing", preprocessor),
    ("classifier", RandomForestClassifier(
        n_estimators=300,
        random_state=42,
        class_weight="balanced"
    ))
])

model.fit(X_train, y_train)
y_pred = model.predict(X_test)

print("Test accuracy:", round(accuracy_score(y_test, y_pred), 4))
print("Macro F1-score:", round(f1_score(y_test, y_pred, average="macro"), 4))
print("\nClassification report:")
print(classification_report(y_test, y_pred))

ConfusionMatrixDisplay.from_predictions(
    y_test, y_pred,
    cmap="Blues",
    values_format="d"
)
plt.title("Random Forest: Cluster Label Prediction")
plt.tight_layout()
plt.show()
```
